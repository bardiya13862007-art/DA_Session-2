```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
cities = ["Tehran" , "Shiraz" , "Esfehan" , "Kashan" , "Tabriz" , "Yazd" , "Mashahd" , "Kerman" , "Rasht" , "Ramsar"]

population = np.array([
    8693706,
    1565572,
    1961260,
    304000,
    1558693,
    529673,
    3001184,
    537718,
    679995,
    35000
])
latitude = np.array([
    35.69439,
    29.61031,
    32.65246,
    33.9850,
    38.08000,
    31.89722,
    36.29807,
    30.28321,
    37.27611,
    36.90336
])

longitude = np.array([
    51.42151,
    52.53113,
    51.67462,
    51.4100,
    46.29190,
    54.36750,
    59.60567,
    57.07879,
    49.58862,
    50.65858
])
```

```python
plt.scatter(longitude, latitude, population/10000)

for i in range(len(cities)):
    plt.text(longitude[i], latitude[i], cities[i])

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Cities Location")

plt.show()
```

```python
plt.bar(cities , population)
plt.xticks(rotation=45)


plt.xlabel("Cities")
plt.ylabel("Population")
plt.title("Cities Population")

plt.show()
```

```python
plt.plot(cities , population)
plt.xticks(rotation=45)


plt.xlabel("Cities")
plt.ylabel("Population")
plt.title("Cities Population")

plt.show()
```

```python
plt.pie(population, labels=cities, autopct='%1.1f%%')

plt.title("Cities Population")

plt.show()
```
